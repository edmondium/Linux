```python
perpustakaan = [('Penulis', 'Judul', 'Halaman'), ('Edmond', 'Belajar Python', 100), ('Budi', 'Belajar Java', 200), ('Siti', 'Belajar C++', 300)]
```

```python
perpustakaan
```

```python
for buku in perpustakaan:
    print(buku)
```

```python
for buku in perpustakaan:
    print(f'Nama penulis adalah {buku[0]}')
```

```python
for penulis, judul, halaman in perpustakaan:
    print(f'Nama penulis adalah {penulis}')
```

```python
for penulis, judul, halaman in perpustakaan:
    print(f' {penulis} {judul} {halaman}')
```

```python
for penulis, judul, halaman in perpustakaan:
    print(f' {penulis:10} {judul:20} {halaman:15}')
```

```python
for penulis, judul, halaman in perpustakaan:
    print(f' {penulis:10} {judul:20} {halaman:>15}')
```

```python
for penulis, judul, halaman in perpustakaan:
    print(f' {penulis:10} {judul:20} {halaman:.>15}')
```
